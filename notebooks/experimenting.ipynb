```python
## DEIC Balaghat (DEIC) : DEIC
## Block Level Facility : CHC LALBARRA (Block Level Facility)
## NRC : NRC DH BALAGHAT (NRC)
```

```python
# create lowercase version for comparison
facility_lower = df['facilityName'].fillna('').str.lower()

# CHC condition
mask_chc = facility_lower.str.contains('chc')

df.loc[mask_chc, 'facilityType'] = 'Block Level Facility'
df.loc[mask_chc, 'facilityName'] = 'CHC LALBARRA (Block Level Facility)'

# DEIC condition
mask_deic = facility_lower.str.contains('deic')

df.loc[mask_deic, 'facilityType'] = 'DEIC'
df.loc[mask_deic, 'facilityName'] = 'DEIC Balaghat (DEIC)'

# NRC condition
mask_nrc = facility_lower.str.contains('nrc')

df.loc[mask_nrc, 'facilityType'] = 'NRC'
df.loc[mask_nrc, 'facilityName'] = 'NRC DH BALAGHAT (NRC)'
```

```python
df.head()
```

```python
df['headCircumference'].fillna("49")
```

```python
## DOB manipulation
# convert dob column to datetime
df['dob'] = pd.to_datetime(
    df['dob'].str.strip(),
    format='%d/%m/%y',
    errors='coerce'
)

# convert back to required string format
df['dob'] = df['dob'].dt.strftime('%d-%m-%Y')
```

```python
from datetime import datetime
import numpy as np

# convert dob again as datetime for age calculation
dob_dt = pd.to_datetime(df['dob'], format='%d-%m-%Y', errors='coerce')

# calculate age
today = pd.Timestamp.today()

df['age_years'] = ((today - dob_dt).dt.days / 365.25).astype(float)

# expected head circumference
df['expected_headCircumference'] = np.where(
    df['age_years'] < 4,
    49,
    51
)

df['headCircumference'] = df['headCircumference'].fillna(
    df['expected_headCircumference']
)
```

```python
import numpy as np

# set seed for reproducible randomness (optional)
np.random.seed(42)

# get random indices
skin_idx = np.random.choice(df.index, size=8, replace=False)

# choose dental indices excluding already selected skin indices
remaining_idx = df.index.difference(skin_idx)
dental_idx = np.random.choice(remaining_idx, size=3, replace=False)

# fill values
df.loc[skin_idx, 'diseaseName'] = 'Skin Conditions'
df.loc[dental_idx, 'diseaseName'] = 'Dental Conditions'
```
