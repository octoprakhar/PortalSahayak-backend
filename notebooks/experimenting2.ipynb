```python
import re

def clean_column_name(col:str):

    # Just remove leading and trailing space
    col = col.strip()

    # if column name is something like 'unnamed' remove it
    if col.lower().startswith('unnamed'):
        return None
    
    ## If column name is already written in correct form then just leave it
    if re.match(r'^[a-z]+([A-Z][a-z0-9]*)*$', col):
        return col
    
    # Removing special characters with a space
    col = re.sub(r'[_\-\s]+', ' ', col)

    ## Making camel case
    words = col.split()

    return words[0].lower() + ''.join(word.capitalize() for word in words[1:])
```

```python
cleaned_columns = [clean_column_name(col) for col in df.columns]

df = df.loc[:, [c is not None for c in cleaned_columns]]

df.columns = [c for c in cleaned_columns if c is not None]
```

### Start by filling headCircumference wrt age
* 0-1 : 43
* 1-2 : 46
* 2-3 : 48
* 3+ : 50

```python
def get_age(dob):
    """
    Returns age in years from dob.
    Accepts string, datetime, Timestamp, or None.
    """
    if pd.isna(dob):
        return None
    
    # Convert to string and remove spaces
    dob_str = str(dob).strip()

    # Require day/month/year
    pattern = r'^\d{1,2}[/-]\d{1,2}[/-]\d{2,4}$'

    if not re.match(pattern, dob_str):
        return None
    
    ## Now let's convert it safely
    dob = pd.to_datetime(
        dob,
        dayfirst=True,
        errors='coerce'
    )

    ## If invalid date
    if pd.isna(dob):
        return None
    
    ## Calculating age
    today = pd.Timestamp.today()

    return (today-dob).days / 365.25
```

```python
def get_head_circumference(age):
    """
    Returns head circumference of the child if age is given
    """
    if pd.isna(age):
        return None
    
    if age < 1:
        return 43
    elif age < 2:
        return 46
    elif age < 3:
        return 48
    else:
        return 50
    

## Only convert those headCircumference which are null
mask = df["headCircumference"].isna()

df.loc[mask,"headCircumference"] = (
    df.loc[mask,'dob']
    .apply(get_age)
    .apply(get_head_circumference)
)
```

### Now transforming dob

```python

def clean_dob(dob):
    """
    For any format of DOB , it changes to dd-mm-yyyy, and string
    """

    if pd.isna(dob):
        return pd.NA
    
    dob_str = str(dob).strip()

    # Require day-month-year structure
    pattern = r'^\d{1,2}[/-]\d{1,2}[/-]\d{2,4}$'

    if not re.match(pattern, dob_str):
        return pd.NA
    
    ## Convert it safely
    dob = pd.to_datetime(
        dob_str,
        dayfirst=True,
        errors='coerce'
    )

    if pd.isna(dob):
        return pd.NA
    
    return dob.strftime('%d-%m-%Y')
```

```python
import numpy as np

def fill_disease_from_nan(df, column='diseaseName', ratio=0.15, seed=42):

    np.random.seed(seed)

    # Step 1: only NaN rows
    nan_indices = df[df[column].isna()].index

    # Step 2: pick 15% of NaNs
    n = int(ratio * len(nan_indices))

    if n == 0:
        return df  # nothing to fill

    selected = np.random.choice(nan_indices, size=n, replace=False)

    # Step 3: split 3:1
    skin_count = int(0.75 * n)

    skin_idx = selected[:skin_count]
    dental_idx = selected[skin_count:]

    # Step 4: fill
    df.loc[skin_idx, column] = 'Skin Conditions'
    df.loc[dental_idx, column] = 'Dental Conditions'

    return df
```

```python
import os
import pandas as pd
from datetime import datetime

class SAMClassifier:
    def __init__(self):

        base_dir = os.path.dirname(os.getcwd())  # folder of SamClassifier.py

        who_dir = os.path.join(base_dir, "WHO_data")

        self.tables = {

            ("male", "wfl"):
                pd.read_excel(

                    os.path.join(
                        who_dir,
                        "wfl_boys_0-to-2-years_zscores.xlsx"
                    )
                ),

            ("female", "wfl"):
                pd.read_excel(
                    os.path.join(
                        who_dir,
                        "wfl_girls_0-to-2-years_zscores.xlsx"
                    )
                ),

            ("male", "wfh"):
                pd.read_excel(
                    os.path.join(
                        who_dir,
                        "wfh_boys_2-to-5-years_zscores.xlsx"
                    )
                ),

            ("female", "wfh"):
                pd.read_excel(
                    os.path.join(
                        who_dir,
                        "wfh_girls_2-to-5-years_zscores.xlsx"
                    )
                )
        }

    def age_months(self,dob:str):
        """
        It converts age in months. And dob must be in 
        string and format dd-mm-yyyy
        """

        if pd.isna(dob):
            return None
    
        dob = datetime.strptime(dob,"%d-%m-%Y")
        days = (datetime.today() - dob).days

        return days / 30.4375
    
    def get_table(self,sex:str,age_months):
        sex = sex.lower()

        if age_months < 24:
            return self.tables[(sex,"wfl")]
        
        return self.tables[(sex, "wfh")]
    
    def get_measurement_column(self,df:pd.DataFrame):
        """
        It will return which column actually the table has
        like either it has 'Height' or 'Length'
        """

        if "Length" in df.columns:
            return "Length"
        if "Height" in df.columns:
            return "Height"
        
        raise Exception("No Length/height column found")
    
    def nearest_row(self, df:pd.DataFrame, measurement_col:str, value):
        """
        It gives the nearest row after substracting value from each measurement col
        so nearest wins. We will get the row from which we will need to calculate the weight.
        """
        idx = (
            df[measurement_col] - value
        ).abs().idxmin()

        return df.loc[idx]
    
    def classify(self, dob, sex, weight, height):

        if pd.isna(dob):
            return None

        age = self.age_months(dob)

        df = self.get_table(sex, age)

        measurement_col = self.get_measurement_column(df)

        row = self.nearest_row(
            df,
            measurement_col,
            height
        )

        minus3sd = row["SD3neg"]
        minus2sd = row["SD2neg"]
        minus1sd = row["SD1neg"]

        # --------------------------------

        if weight < minus2sd:

            status = "SAM"
            message = "Severe Acute Malnutrition"

        elif weight < minus1sd:

            status = "MAM"
            message = "Moderate Acute Malnutrition"

        else:

            status = "Normal"
            message = "No acute malnutrition"

        return {

            "age_months": round(age, 1),

            "used_standard":
                "WFL" if age < 24 else "WFH",

            "minus3sd_cutoff":
                minus3sd,

            "minus2sd_cutoff":
                minus2sd,

            "status":
                status,

            "message":
                message
        }
```

```python
def standardize_facility(df, classifier):

    # -----------------------------------
    # CLEAN FACILITY COLUMN
    # -----------------------------------

    facility_clean = (
        df['facilityName']
        .fillna('')
        .astype(str)
        .str.strip()
    )

    facility_lower = facility_clean.str.lower()

    # -----------------------------------
    # PROCESS EACH ROW
    # -----------------------------------

    for idx, row in df.iterrows():

        if pd.isna(row['dob']):
            continue

        facility = str(row['facilityName']).strip().lower()

        # -------------------------------
        # CHECK SAM STATUS
        # -------------------------------

        result = classifier.classify(

            dob=row['dob'],

            sex=row['gender'],

            weight=float(row['weight']),

            height=float(row['height'])
        )

        is_sam = result["status"] == "SAM"

        # -------------------------------
        # EMPTY FACILITY
        # -------------------------------

        if facility == '' or facility == 'nan':

            if is_sam:

                df.loc[idx, 'facilityType'] = 'NRC'

                df.loc[idx, 'facilityName'] = (
                    'NRC DH BALAGHAT (NRC)'
                )

            else:

                df.loc[idx, 'facilityType'] = (
                    'Block Level Facility'
                )

                df.loc[idx, 'facilityName'] = (
                    'CHC LALBARRA (Block Level Facility)'
                )

        # -------------------------------
        # CHC OR NRC WRITTEN
        # -------------------------------

        elif ('chc' in facility) or ('nrc' in facility):

            if is_sam:

                df.loc[idx, 'facilityType'] = 'NRC'

                df.loc[idx, 'facilityName'] = (
                    'NRC DH BALAGHAT (NRC)'
                )

            else:

                df.loc[idx, 'facilityType'] = (
                    'Block Level Facility'
                )

                df.loc[idx, 'facilityName'] = (
                    'CHC LALBARRA (Block Level Facility)'
                )

        # -------------------------------
        # DEIC CONDITION
        # -------------------------------

        elif 'deic' in facility:

            df.loc[idx, 'facilityType'] = 'DEIC'

            df.loc[idx, 'facilityName'] = (
                'DEIC Balaghat (DEIC)'
            )

    return df

classfier = SAMClassifier()
df = standardize_facility(df,classifier=classfier)
df.head()
```

## If N number of SAM found and it's facility is either of these (CHC, NRC, or NONE) then just modify height and weight like this(including decimals) and write facilityType and facilityName for CHC one :
* Age 0-1 = weight: 6-8, height: 60-70
* Age 1-2 = weight : 9-11 , height: 70-80
* Age 2-3 = weight : 9-11 , height : 80-90
* Age 3-4 = weight: 12-14, height : 90-100
* Age 4-5 = weight: 13-15, height: 95-105
* Age 5+ = weight: 14-16, height: 100-110

```python
import random
import pandas as pd


def get_corrected_values(age_years):
    """
    Returns corrected weight and height
    according to age group.
    """

    if age_years < 1:
        return (
            round(random.uniform(6, 8), 1),
            round(random.uniform(60, 70), 1)
        )

    elif age_years < 2:
        return (
            round(random.uniform(9, 11), 1),
            round(random.uniform(70, 80), 1)
        )

    elif age_years < 3:
        return (
            round(random.uniform(9, 11), 1),
            round(random.uniform(80, 90), 1)
        )

    elif age_years < 4:
        return (
            round(random.uniform(12, 14), 1),
            round(random.uniform(90, 100), 1)
        )

    elif age_years < 5:
        return (
            round(random.uniform(13, 15), 1),
            round(random.uniform(95, 105), 1)
        )

    else:
        return (
            round(random.uniform(14, 16), 1),
            round(random.uniform(100, 110), 1)
        )
```

```python
def standardize_facility_2(df, classifier, n=3):

    sam_count = 0

    for idx, row in df.iterrows():

        # --------------------------------
        # SKIP INVALID DOB
        # --------------------------------

        if pd.isna(row['dob']):
            continue

        facility = str(row['facilityName']).strip().lower()

        # --------------------------------
        # CLASSIFY CHILD
        # --------------------------------

        result = classifier.classify(

            dob=row['dob'],

            sex=row['gender'],

            weight=float(row['weight']),

            height=float(row['height'])
        )

        is_sam = result["status"] == "SAM"

        # --------------------------------
        # FACILITY TYPES TO HANDLE
        # --------------------------------

        valid_facility = (
            facility == ''
            or facility == 'nan'
            or 'chc' in facility
            or 'nrc' in facility
        )

        # --------------------------------
        # IF SAM FOUND
        # --------------------------------

        if is_sam and valid_facility:

            # FIRST N SAM CHILDREN -> NRC
            if sam_count < n:

                df.loc[idx, 'facilityType'] = 'NRC'

                df.loc[idx, 'facilityName'] = (
                    'NRC DH BALAGHAT (NRC)'
                )

                sam_count += 1

            # REMAINING SAM -> MODIFY VALUES + CHC
            else:

                age_years = result["age_months"] / 12

                new_weight, new_height = (
                    get_corrected_values(age_years)
                )

                # MODIFY VALUES
                df.loc[idx, 'weight'] = new_weight

                df.loc[idx, 'height'] = new_height

                # SEND TO CHC
                df.loc[idx, 'facilityType'] = (
                    'Block Level Facility'
                )

                df.loc[idx, 'facilityName'] = (
                    'CHC LALBARRA (Block Level Facility)'
                )

        # --------------------------------
        # DEIC CONDITION
        # --------------------------------

        elif 'deic' in facility:

            df.loc[idx, 'facilityType'] = 'DEIC'

            df.loc[idx, 'facilityName'] = (
                'DEIC Balaghat (DEIC)'
            )

        # --------------------------------
        # NORMAL CHILDREN
        # --------------------------------

        else:

            df.loc[idx, 'facilityType'] = (
                'Block Level Facility'
            )

            df.loc[idx, 'facilityName'] = (
                'CHC LALBARRA (Block Level Facility)'
            )

    return df
```
